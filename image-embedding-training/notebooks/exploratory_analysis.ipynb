# Exploratory Data Analysis

In this notebook, we will perform exploratory data analysis (EDA) on the image dataset used for training the embedding model. The goal is to visualize the dataset and understand the distribution of images.

```python
# Import necessary libraries
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

# Set visualization style
sns.set(style='whitegrid')
```

```python
# Load image data
image_dir = '../data/raw/'  # Adjust path as necessary
image_files = [f for f in os.listdir(image_dir) if f.endswith(('.png', '.jpg', '.jpeg'))]
num_images = len(image_files)

print(f'Total number of images: {num_images}')
```

```python
# Visualize a sample of images
sample_images = np.random.choice(image_files, size=9, replace=False)

plt.figure(figsize=(10, 10))
for i, img_file in enumerate(sample_images):
    img = Image.open(os.path.join(image_dir, img_file))
    plt.subplot(3, 3, i + 1)
    plt.imshow(img)
    plt.axis('off')
    plt.title(img_file)
plt.tight_layout()
plt.show()
```

```python
# Analyze image dimensions
image_dimensions = []

for img_file in image_files:
    img = Image.open(os.path.join(image_dir, img_file))
    image_dimensions.append(img.size)

widths, heights = zip(*image_dimensions)
plt.figure(figsize=(12, 6))
sns.histplot(widths, bins=30, color='blue', label='Width', kde=True)
sns.histplot(heights, bins=30, color='orange', label='Height', kde=True)
plt.legend()
plt.title('Distribution of Image Dimensions')
plt.xlabel('Pixels')
plt.ylabel('Frequency')
plt.show()
```
